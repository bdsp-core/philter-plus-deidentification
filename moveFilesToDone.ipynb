```python
import pandas as pd
import os
import shutil

# Constants for the directory paths
WAITING_ROOM_DIR = 'data/waitingRoom'
DONE_FOLDER_DIR = 'data/doneFolder'
CSV_FILE_PATH = 'schedulingCSV.csv'

# Ensure the done folder exists
os.makedirs(DONE_FOLDER_DIR, exist_ok=True)

# Read the CSV file into a DataFrame
df = pd.read_csv(CSV_FILE_PATH)

```

```python
df
```

```python
# Filter the DataFrame for rows where `done` is marked as 1
done_files = df[df['done'] == 1]['inputFileName']

# Loop through the filenames and move them to the done folder
for file_name in done_files:
    source_path = os.path.join(WAITING_ROOM_DIR, file_name)
    destination_path = os.path.join(DONE_FOLDER_DIR, file_name)
    
    # Check if the file exists in the waiting room
    if os.path.exists(source_path):
        # Move the file
        shutil.move(source_path, destination_path)
        print(f"Moved file {file_name} to {DONE_FOLDER_DIR}")
    else:
        print(f"File {file_name} not found in {WAITING_ROOM_DIR}")
```
