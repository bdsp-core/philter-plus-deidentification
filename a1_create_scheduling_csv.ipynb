```python
# create a list of files in the waiting room
# columns: file name, done (i.e. already in output), batch number
waitingRoom = "/Users/mbw/cdac Dropbox/brandon westover/ECG_Deidentification/philter-plus-deidentification/data/waitingRoom"
outputFolder = "/Users/mbw/cdac Dropbox/brandon westover/ECG_Deidentification/philter-plus-deidentification/data/output"

import os
import pandas as pd

import os
import pandas as pd

N = 10 # Total number of batches

df = pd.DataFrame(columns=['inputFileName', 'outputFileName', 'done', 'batchNumber'])

n_files = len(os.listdir(waitingRoom))
batch_size = n_files // N 

batch_num = 1
for i, f in enumerate(os.listdir(waitingRoom)):
  if i % batch_size == 0 and i > 0:
    batch_num += 1
    
  new_row = {'inputFileName': f, 'batchNumber': batch_num}
  df = pd.concat([df, pd.DataFrame([new_row])], ignore_index=True) 

for f in os.listdir(outputFolder):
  if f in df['inputFileName'].values:
    idx = df.loc[df['inputFileName'] == f].index[0]
    df.at[idx, 'outputFileName'] = f
    df.at[idx, 'done'] = 1
    
print(df)
```

```python
df['newOutputFileName'] = df['inputFileName'].str.replace('input', 'deID')
```
