```python
import os

chunk_num = 1
chunk_size = 1000

with open('source/ecg_diagnosis_input_2022.txt') as f:
    chunk = []
    print(chunk_num)
    for line in f:
        chunk.append(line)
        if len(chunk) == chunk_size:
            out_file = f'waitingRoom/chunk{chunk_num}.txt'
            with open(out_file, 'w') as out: 
                out.writelines(chunk)
            chunk_num += 1
            chunk = []

    if chunk:
        out_file = f'input/chunk{chunk_num}.txt'
        with open(out_file, 'w') as out:
            out.writelines(chunk)
```
