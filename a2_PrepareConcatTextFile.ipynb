```python
import os
import pandas as pd

input_folder = '/Users/mbw/cdac Dropbox/brandon westover/ecg_deidentification_input'
output_dir = '/Users/mbw/cdac Dropbox/brandon westover/ECG_Deidentification/philter-plus-deidentification/data/waitingRoom'
csv_files = [f for f in os.listdir(input_folder) if f.endswith('.csv')]

def fcnChunkify(source_csv):
    df = pd.read_csv(os.path.join(input_folder, source_csv))
    N = 1000
    separator = r"------------------------------------------------------------------------------------------------"
    chunk_id = 1

    # Iterate through DataFrame in chunks of N rows
    for start_row in range(0, len(df), N):
        end_row = min(start_row + N, len(df))  # Ensure we don't go past the end of the DataFrame
        chunk = df.iloc[start_row:end_row]  # Define the chunk

        # Create or overwrite a chunk file
        chunk_file = f"{source_csv[:-4]}_chunk{chunk_id}.txt"
        with open(os.path.join(output_dir, chunk_file), 'w') as f:  # 'w' mode to overwrite existing file
            for index, row in chunk.iterrows():
                diagnosis = str(row['DiagnosisTXT'])
                if diagnosis == 'nan':
                    diagnosis = ''
                f.write(diagnosis + "\n")
                f.write(separator + "\n")
        chunk_id += 1
    
    # Move processed source CSV
    source_csv_path = os.path.join(input_folder, source_csv)  
    new_folder = os.path.join(input_folder, 'AlreadySplitInChunks')
    if not os.path.exists(new_folder):
        os.makedirs(new_folder)
    os.rename(source_csv_path, os.path.join(new_folder, source_csv))
```

```python
for i in range(len(csv_files)):
    source_csv = os.path.basename(csv_files[i])
    print(source_csv)
    fcnChunkify(source_csv)
```
